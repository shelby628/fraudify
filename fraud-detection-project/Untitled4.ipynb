```python
import requests
import json

url = "http://127.0.0.1:8000/api/predict/"

token = "d67a2e002528aca932f2c5ae42c5d895bfd63d6e"  # <-- REQUIRED

headers = {
    "Authorization": f"Token {token}",
    "Content-Type": "application/json"
}

transaction_data = {
    "step": 13,
    "amount": 600000,
    "oldbalanceOrg": 600000,
    "newbalanceOrig": 0,
    "oldbalanceDest": 0,
    "newbalanceDest": 6000,
    "type_CASH_OUT": 1,
    "type_DEBIT": 0,
    "type_PAYMENT": 0,
    "type_TRANSFER": 0
}

response = requests.post(
    url,
    headers=headers,
    json=transaction_data   # 👈 use json=, not data=
)

print(response.status_code)
print(response.json())
```
