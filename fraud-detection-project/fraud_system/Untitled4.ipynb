```python
from django.contrib.auth.models import User
from rest_framework.authtoken.models import Token

# Choose an existing user
user = User.objects.get(username="regular_user")  # or your username

# Get or create a token
token, created = Token.objects.get_or_create(user=user)
print("Your token:", token.key)
```

```python
import requests
import json

url = "http://127.0.0.1:8000/api/predict/"

payload = {
    "step": 13,
    "amount": 600000,
    "oldbalanceOrg": 600000,
    "newbalanceOrig": 0,
    "oldbalanceDest": 0,
    "newbalanceDest": 6000,
    "type_CASH_OUT": 1,
    "type_DEBIT": 0,
    "type_PAYMENT": 0,
    "type_TRANSFER": 0
}

headers = {
    "Authorization": "Token d67a2e002528aca932f2c5ae42c5d895bfd63d6e",  # replace with your token
    "Content-Type": "application/json"
}

r = requests.post(url, headers=headers, data=json.dumps(payload))
print(r.status_code)
print(r.json())
```
